```python
import pandas as pd

# Ensure the dataset is loaded and cleaned
df = pd.read_csv('/content/WA_Fn-UseC_-Telco-Customer-Churn.csv.csv')
df_clean = df.copy()
df_clean = df_clean.apply(lambda col: col.str.strip() if col.dtype == 'object' else col)
df_clean['TotalCharges'] = pd.to_numeric(df_clean['TotalCharges'], errors='coerce')

# Create the crosstab
contract_table = pd.crosstab(
    df_clean['Contract'],
    df_clean['Churn']
)

contract_table
```

```python
chi2, p_value, dof, expected = chi2_contingency(contract_table)

print("Chi-square statistic:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", dof)
```

```python
if p_value < 0.05:
    print("Reject the null hypothesis.")
    print("There is a statistically significant association.")
else:
    print("Fail to reject the null hypothesis.")
    print("There is insufficient evidence of an association.")
```

```python
internet_table = pd.crosstab(
    df_clean['InternetService'],
    df_clean['Churn']
)

internet_table
```

```python
chi2, p_value, dof, expected = chi2_contingency(internet_table)

print("Chi-square statistic:", chi2)
print("P-value:", p_value)

if p_value < 0.05:
    print("There is a statistically significant association.")
else:
    print("There is insufficient evidence of an association.")
```

```python
chi2, p_value, dof, expected = chi2_contingency(internet_table)

print("Chi-square statistic:", chi2)
print("P-value:", p_value)

if p_value < 0.05:
    print("There is a statistically significant association.")
else:
    print("There is insufficient evidence of an association.")
```

```python
def evaluate_model(model_name, y_true, y_pred, y_prob):

    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall = recall_score(y_true, y_pred, zero_division=0)
    f1 = f1_score(y_true, y_pred, zero_division=0)
    roc_auc = roc_auc_score(y_true, y_prob)

    return {
        'Model': model_name,
        'Accuracy': accuracy,
        'Precision': precision,
        'Recall': recall,
        'F1 Score': f1,
        'ROC-AUC': roc_auc
    }
```

Calculate evaluation results

```python
import pandas as pd
import os
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

# 1. Reconstruct variables if they are missing from the active runtime state
if 'X_test' not in globals() or 'y_test' not in globals() or 'logistic_model' not in globals():
    # Order of path checks based on your notebook history
    possible_paths = [
        '/content/WA_Fn-UseC_-Telco-Customer-Churn.csv.csv',
        '/content/WA_Fn-UseC_-Telco-Customer-Churn.csv'
    ]

    file_path = None
    for path in possible_paths:
        if os.path.exists(path):
            file_path = path
            break

    # If still not present, download the dataset from a verified, stable public source
    if not file_path:
        file_path = '/content/WA_Fn-UseC_-Telco-Customer-Churn.csv.csv'
        print("Dataset not found in runtime. Downloading dataset from a stable public repository...")
        import urllib.request
        # Verified stable URL
        url = "https://raw.githubusercontent.com/shubhampatodi/telco-customer-churn-prediction/master/WA_Fn-UseC_-Telco-Customer-Churn.csv"
        urllib.request.urlretrieve(url, file_path)

    # Load and preprocess data
    df = pd.read_csv(file_path)
    df_clean = df.copy()
    df_clean = df_clean.apply(lambda col: col.str.strip() if col.dtype == 'object' else col)
    df_clean['TotalCharges'] = pd.to_numeric(df_clean['TotalCharges'], errors='coerce')

    df_model = df_clean.drop(columns=['customerID'])
    df_model['Churn'] = df_model['Churn'].map({'Yes': 1, 'No': 0})

    X = df_model.drop(columns=['Churn'])
    y = df_model['Churn']

    # Features
    numeric_features = ['SeniorCitizen', 'tenure', 'MonthlyCharges', 'TotalCharges']
    categorical_features = [col for col in X.columns if col not in numeric_features]

    from sklearn.impute import SimpleImputer
    from sklearn.preprocessing import OneHotEncoder, StandardScaler
    from sklearn.compose import ColumnTransformer
    from sklearn.pipeline import Pipeline

    numeric_transformer = Pipeline(steps=[
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ])

    categorical_transformer = Pipeline(steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('onehot', OneHotEncoder(handle_unknown='ignore'))
    ])

    preprocessor = ColumnTransformer(transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ])

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

    # Train Models
    logistic_model = Pipeline(steps=[('preprocessor', preprocessor), ('classifier', LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42))])
    decision_tree_model = Pipeline(steps=[('preprocessor', preprocessor), ('classifier', DecisionTreeClassifier(max_depth=5, class_weight='balanced', random_state=42))])
    random_forest_model = Pipeline(steps=[('preprocessor', preprocessor), ('classifier', RandomForestClassifier(n_estimators=200, max_depth=10, class_weight='balanced', random_state=42, n_jobs=-1))])

    logistic_model.fit(X_train, y_train)
    decision_tree_model.fit(X_train, y_train)
    random_forest_model.fit(X_train, y_train)

    y_pred_logistic = logistic_model.predict(X_test)
    y_pred_tree = decision_tree_model.predict(X_test)
    y_pred_forest = random_forest_model.predict(X_test)

# 2. Model Evaluation and Performance comparison
results = []

def evaluate_model(model_name, y_true, y_pred, y_prob):
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall = recall_score(y_true, y_pred, zero_division=0)
    f1 = f1_score(y_true, y_pred, zero_division=0)
    roc_auc = roc_auc_score(y_true, y_prob)

    return {
        'Model': model_name,
        'Accuracy': accuracy,
        'Precision': precision,
        'Recall': recall,
        'F1 Score': f1,
        'ROC-AUC': roc_auc
    }

results.append(
    evaluate_model(
        'Logistic Regression',
        y_test,
        y_pred_logistic,
        logistic_model.predict_proba(X_test)[:, 1]
    )
)

results.append(
    evaluate_model(
        'Decision Tree',
        y_test,
        y_pred_tree,
        decision_tree_model.predict_proba(X_test)[:, 1]
    )
)

results.append(
    evaluate_model(
        'Random Forest',
        y_test,
        y_pred_forest,
        random_forest_model.predict_proba(X_test)[:, 1]
    )
)

results_df = pd.DataFrame(results)
results_df
```

```python
import pandas as pd
import os
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

# Automatically reconstruct 'rf_importance_df' if it is missing from the active session
if 'rf_importance_df' not in globals():
    print("Rebuilding RandomForest model and feature importances to define 'rf_importance_df'...")

    # Load and clean dataset
    file_path = '/content/WA_Fn-UseC_-Telco-Customer-Churn.csv.csv'
    if not os.path.exists(file_path):
        file_path = '/content/WA_Fn-UseC_-Telco-Customer-Churn.csv'

    df = pd.read_csv(file_path)
    df_clean = df.copy()
    df_clean = df_clean.apply(lambda col: col.str.strip() if col.dtype == 'object' else col)
    df_clean['TotalCharges'] = pd.to_numeric(df_clean['TotalCharges'], errors='coerce')

    df_model = df_clean.drop(columns=['customerID'])
    df_model['Churn'] = df_model['Churn'].map({'Yes': 1, 'No': 0})

    X = df_model.drop(columns=['Churn'])
    y = df_model['Churn']

    numeric_features = ['SeniorCitizen', 'tenure', 'MonthlyCharges', 'TotalCharges']
    categorical_features = [col for col in X.columns if col not in numeric_features]

    from sklearn.impute import SimpleImputer
    from sklearn.preprocessing import OneHotEncoder, StandardScaler
    from sklearn.compose import ColumnTransformer
    from sklearn.pipeline import Pipeline

    numeric_transformer = Pipeline(steps=[
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ])

    categorical_transformer = Pipeline(steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('onehot', OneHotEncoder(handle_unknown='ignore'))
    ])

    preprocessor = ColumnTransformer(transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ])

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

    # Train Random Forest
    random_forest_model = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('classifier', RandomForestClassifier(n_estimators=200, max_depth=10, class_weight='balanced', random_state=42, n_jobs=-1))
    ])
    random_forest_model.fit(X_train, y_train)

    # Extract feature names and importance score
    feature_names = random_forest_model.named_steps['preprocessor'].get_feature_names_out()
    rf_importance = random_forest_model.named_steps['classifier'].feature_importances_

    rf_importance_df = pd.DataFrame({
        'Feature': feature_names,
        'Importance': rf_importance
    })
    rf_importance_df = rf_importance_df.sort_values(by='Importance', ascending=False)

rf_importance_df.head(10)
```

Predict churn for individual customers

```python
customer_predictions = X_test.copy()

customer_predictions['Actual_Churn'] = y_test

customer_predictions['Predicted_Churn'] = (
    random_forest_model.predict(X_test)
)

customer_predictions['Churn_Probability'] = (
    random_forest_model.predict_proba(X_test)[:, 1]
)

customer_predictions.head(10)
```

High risk customers

```python
high_risk_customers = customer_predictions[
    customer_predictions['Churn_Probability'] >= 0.70
]

high_risk_customers.head(10)
```

count of high risk customers

```python
print(
    "Number of high-risk customers:",
    len(high_risk_customers)
)
```

```python
print(
    "Percentage of high-risk customers:",
    round(
        len(high_risk_customers) / len(customer_predictions) * 100,
        2
    ),
    "%"
)
```

visualize churn risk

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.histplot(
    customer_predictions['Churn_Probability'],
    bins=20,
    kde=True
)

plt.title('Distribution of Predicted Customer Churn Risk')
plt.xlabel('Predicted Churn Probability')
plt.ylabel('Number of Customers')

plt.show()
```
